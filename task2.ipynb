```python
sentence = "coding is easy and coding is powerful"

words = sentence.split() 
unique_words = set(words)

print("Unique words count:", len(unique_words))
print("Unique words:", unique_words) 
```

```python
employees = {
    "antony": 45000,
    "eloisee": 128000,
    "sophie": 77000
}

highest_employee = max(employees, key=employees.get) 

print("Highest Salary:", highest_employee, "-", employees[highest_employee]) 

```

```python
numbers = [85,92, 29, 15, 44]
max_value = numbers[0]
min_value = numbers[0]

for num in numbers:
    if num > max_value:
        max_value = num
    if num < min_value:
        min_value = num

print("List:", numbers)
print("Max:", max_value)
print("Min:", min_value)
```

```python
prices = [677,2026, 5698, 676, 900]
count = 0

for price in prices:
    if price > 1000:
        count += 1

print("Products above 1000:", count) 
```

```python
attendance = ["P", "A", "A", "P", "A"]

present_count = attendance.count("P") 
total_days = len(attendance)

attendance_percentage = (present_count / total_days) * 100 
print("Attendance Percentage:", attendance_percentage)
```

```python
text = "Lannister"
char_count = {}

for char in text:
    if char in char_count:
        char_count[char] += 1
    else:
        char_count[char] = 1

print(char_count) 
```
