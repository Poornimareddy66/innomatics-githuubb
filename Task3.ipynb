```python
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

highest_score = max(employees.values())
top_employees = []

for name in employees:
    if employees[name] == highest_score:
        top_employees.append(name)


print("Top Performers Eligible for Bonus:", ", ".join(top_employees), "(Score:", highest_score, ")")
```

```python
query = "Buy mobile phone buy phone online"
query = query.lower()

for ch in ",.!?":
    query = query.replace(ch, "")
words = query.split()
word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1
result = {}

for word in word_count:
    if word_count[word] > 1:
        result[word] = word_count[word]

print(result)
```

```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]

valid_readings = []

for index in range(len(sensor_readings)):
    if sensor_readings[index] % 2 == 0:   # check if even
        valid_readings.append((index, sensor_readings[index]))

print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

```python
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

domain_count = {}

for email in emails:
    domain = email.split("@")[1]
    
    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1

total_emails = len(emails)

for domain in domain_count:
    percentage = (domain_count[domain] / total_emails) * 100
    print(f"{domain}: {int(percentage)}%")
```

```python
sales = [1200, 1500, 900, 2200, 1400, 3000]
average_sales = sum(sales) / len(sales)

threshold = average_sales * 1.30

for i in range(len(sales)):
    if sales[i] > threshold:
        print(f"Day {i+1}: {sales[i]}")
```
