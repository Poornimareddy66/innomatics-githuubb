```python
def smart_parking(capacity, logs):
    parked = 0
    peak = 0
    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1
        peak = max(peak, parked)
    
    status = "Available" if parked < capacity else "Full"
    alert = "Alert: Capacity Exceeded!" if parked > capacity else "No Alert"
    
    return f"Currently Parked Vehicles: {parked}\nParking Status: {status}\nPeak Usage: {peak}\n{alert}"

print(smart_parking(50, ["IN","IN","IN","OUT","IN","IN","OUT"]))
```

```python
def delivery_time(distance, traffic, weather):
    base_time = distance * 5  # 5 min per km
    
    if traffic == "High":
        base_time += 15
    elif traffic == "Medium":
        base_time += 10
    
    if weather == "Rainy":
        base_time += 10
    elif weather == "Stormy":
        base_time += 20
    
    return f"Estimated Delivery Time: {base_time} minutes"

print(delivery_time(8, "High", "Rainy"))
```

```python
def theatre_occupancy(total_seats, booked_seats):
    occupied = len(booked_seats)
    occupancy = (occupied / total_seats) * 100
    
    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Available"
    
    return f"Occupancy: {int(occupancy)}%\nShow Status: {status}"

print(theatre_occupancy(200, [1]*150))
```

```python
def server_load(cpu_readings):
    avg = sum(cpu_readings) / len(cpu_readings)
    
    if avg < 50:
        status = "Normal"
    elif avg <= 80:
        status = "Warning"
    else:
        status = "Critical"
    
    return f"Average CPU Load: {int(avg)}%\nServer Status: {status}"

print(server_load([45, 60, 70, 85, 90]))
```

```python
def resource_monitor(usage):
    overused = [res for res, hrs in usage.items() if hrs > 8]
    alert = "Yes" if overused else "No"
    
    return f"Overused Resources: {', '.join(overused) if overused else 'None'}\nEnergy Alert: {alert}"

print(resource_monitor({"Projector":6,"AC":9,"Lights":4}))
```

```python
def event_registration(capacity, registrations):
    confirmed = min(capacity, registrations)
    waitlist = max(0, registrations - capacity)
    status = "Closed" if registrations >= capacity else "Open"
    
    return f"Confirmed Registrations: {confirmed}\nWaitlisted Users: {waitlist}\nRegistration Status: {status}"

print(event_registration(100, 105))
```
