```python
def validate_recharge(amount):
    valid_plans = [199, 299, 399, 599]

    if amount < 50:
        print("Recharge amount must be at least ₹50.")
        return False

    if amount in valid_plans:
        return True
    else:
        print("Invalid plan selected.")
        return False

while True:
    amount = int(input("Enter recharge amount: ₹"))

    if validate_recharge(amount):
        print(f"Recharge successful for ₹{amount}")
        break
    else:
        print("Please try again.\n")
```

```python
def check_inventory(products):
    for product, stock in products.items():
        if stock < 15:
            print(f"{product} → Reorder Alert (Stock: {stock})")
        else:
            print(f"{product} → Stock OK (Stock: {stock})")

inventory = {
    "Rice": 20,
    "Oil": 10,
    "Sugar": 25,
    "Salt": 12
}
check_inventory(inventory)
```

```python
def process_result(marks):
    total = 0
    

    for mark in marks:
        total += mark

    average = total / len(marks)

    if average >= 50:
        return f"Average: {average:.2f} → Pass"
    else:
        return f"Average: {average:.2f} → Fail"

student_marks = [60, 45, 70, 55]

result = process_result(student_marks)
print(result)
```

```python
def calculate_fare(distance, is_peak):
    base_fare = 50
    per_km_rate = 12
    
    fare = base_fare + (distance * per_km_rate)
    
    if is_peak:
        fare += fare * 0.25   # 25% extra
    
    return fare

while True:
    distance = float(input("Enter distance in km: "))
    peak_input = input("Is it peak hour? (yes/no): ").lower()
    
    is_peak = True if peak_input == "yes" else False
    
    total_fare = calculate_fare(distance, is_peak)
    
    print(f"Total Fare: ₹{total_fare:.2f}")
    
    retry = input("Do you want to calculate again? (yes/no): ").lower()
    if retry != "yes":
        print("Thank you for using Cab Fare Estimator.")
        break
```

```python
def check_password(password):
    if len(password) < 8:
        return "Weak Password: Minimum length should be 8."

    has_digit = False
    has_special = False

    for char in password:
        if char.isdigit():
            has_digit = True
        if char in "@#$":
            has_special = True

    if has_digit and has_special:
        return "Strong Password ✅"
    else:
        return "Weak Password: Must contain at least one digit and one special character (@ # $)."

password = input("Enter your password: ")
result = check_password(password)
print(result)
```
