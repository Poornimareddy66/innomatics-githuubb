```python
def analyze_post_engagement(likes_list):
    total_likes = 0
    
    for likes in likes_list:
        total_likes += likes
    
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    
    print("Total Likes:", total_likes)
    print("Post Status:", status)

likes_data = [200, 300, 150, 400]  
analyze_post_engagement(likes_data)
```

```python
def check_medicine_stock(stock):
    
    if stock < 10:
        status = "Low Stock Alert"
    else:
        status = "Stock Sufficient"
    
    print("Medicine Stock:", stock)
    print("Status:", status)
check_medicine_stock(15)
```

```python
def check_rainfall(rainfall_data, required_level):
    total_rainfall = 0
    
    for rain in rainfall_data:
        total_rainfall += rain
    
    average_rainfall = total_rainfall / len(rainfall_data)
    
    if average_rainfall >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"
    
    print("Average Rainfall:", int(average_rainfall))
    print("Rainfall Status:", status)

season_data = [70, 75, 80, 63]  
required = 70

check_rainfall(season_data, required)
```

```python
def detect_duplicate_accounts(usernames):
    unique_users = set(usernames)
    
    if len(unique_users) < len(usernames):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")

users = ["john", "alice", "john", "mike"]
detect_duplicate_accounts(users)
```

```python
def filter_premium_crops(prices):
    premium = []
    
    for price in prices:
        if price > 2000:
            premium.append(price)
    
    print("Premium Crops:", premium)

crop_prices = [1500, 2500, 1800, 3200]
filter_premium_crops(crop_prices)
```

```python
def check_application_health(errors):
    
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"
    
    print("Error Count:", errors)
    print("System Status:", status)

check_application_health(7)
```

```python
def check_transaction(amount):
    limit = 50000
    
    if amount <= limit:
        status = "Approved"
    else:
        status = "Rejected"
    
    print("Transaction Amount:", amount)
    print("Transaction Status:", status)

check_transaction(60000)
```

```python
def check_attendance(attendance_list):
    total_classes = len(attendance_list)
    attended = 0
    
    for status in attendance_list:
        if status == 1:   # 1 means present
            attended += 1
    
    percentage = (attended / total_classes) * 100
    
    if percentage >= 75:
        eligibility = "Eligible"
    else:
        eligibility = "Not Eligible"
    
    print("Attendance Percentage:", percentage)
    print("Exam Eligibility:", eligibility)

attendance_data = [1,1,1,0,1]  
check_attendance(attendance_data)
```

```python
def calculate_electricity_bill(units):
    bill = 0
    
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)
    
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"
    
    return bill, usage

total_bill, usage_status = calculate_electricity_bill(250)
print("Total Bill:", total_bill)
print("Usage Status:", usage_status)
```
